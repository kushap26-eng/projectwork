```python
import numpy as np


class DataAnalyticcs:

    def __init__(self):
        self.__array = None

    
    @classmethod
    def create_analyzer(cls):
        return cls()

    
    @staticmethod
    def print_line():
        print("=" * 50)

    
    def __check_array(self):
        if self.__array is None:
            print("Please create an array first.")
            return False
        return True

    
    def create_array(self):

        self.print_line()
        print("Create Numpy Array")
        self.print_line()

        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:

                n = int(input("Enter number of elements: "))

                values = list(map(float, input("Enter elements separated by space: ").split()))

                if len(values) != n:
                    print("Invalid number of elements.")
                    return

                self.__array = np.array(values)

            elif choice == 2:

                rows = int(input("Enter number of rows: "))
                columns = int(input("Enter number of columns: "))

                total = rows * columns

                values = list(map(float, input("Enter " + str(total) +" elements separated by space: ").split()))

                if len(values) != total:
                    print("Invalid number of elements.")
                    return

                self.__array = np.array(values).reshape(rows, columns)

            elif choice == 3:

                depth = int(input("Enter depth: "))
                rows = int(input("Enter number of rows: "))
                columns = int(input("Enter number of columns: "))

                total = depth * rows * columns

                values = list(map(float, input("Enter " + str(total) +" elements separated by space: ").split()))

                if len(values) != total:
                    print("Invalid number of elements.")
                    return

                self.__array = np.array(values).reshape(depth, rows, columns)

            else:
                print("Invalid choice.")
                return

            print("Array created successfully:")
            print(self.__array)

        except ValueError:
            print("Please enter valid numeric values.")

    
    def indexing_slicing(self):

        if not self.__check_array():
            return

        self.print_line()
        print("Indexing and Slicing")
        self.print_line()

        print("1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        try:
            choice = int(input("Enter your choice: "))

            if choice == 1:

                try:

                    if self.__array.ndim == 1:

                        index = int(input("Enter index: "))

                        print("Element:",self.__array[index])

                    elif self.__array.ndim == 2:

                        row = int(input("Enter row index: "))
                        column = int(input("Enter column index: "))

                        print("Element:",self.__array[row, column])

                    elif self.__array.ndim == 3:

                        depth = int(input("Enter depth index: "))
                        row = int(input("Enter row index: "))
                        column = int(input("Enter column index: "))

                        print("Element:",self.__array[depth, row, column])

                except (IndexError, ValueError):
                    print("Invalid index.")

            elif choice == 2:

                try:

                    if self.__array.ndim == 1:

                        start = int(input("Enter start index: "))
                        end = int(input("Enter end index: "))

                        print("\nSliced Array:")
                        print(self.__array[start:end])

                    elif self.__array.ndim == 2:

                        row_start = int(input("Enter row start: "))
                        row_end = int(input("Enter row end: "))

                        column_start = int(input("Enter column start: "))
                        column_end = int(input("Enter column end: "))

                        print("\nSliced Array:")
                        print(self.__array[row_start:row_end,column_start:column_end])

                    elif self.__array.ndim == 3:

                        depth_start = int(input("Enter depth start: "))
                        depth_end = int(input("Enter depth end: "))

                        row_start = int(input("Enter row start: "))
                        row_end = int(input("Enter row end: "))

                        column_start = int(input("Enter column start: "))
                        column_end = int(input("Enter column end: "))

                        print("\nSliced Array:")
                        print(self.__array[depth_start:depth_end,row_start:row_end,column_start:column_end])

                except (ValueError, IndexError):
                    print("Invalid input.")

            elif choice == 3:
                return

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter a valid choice.")

    
    def mathematical_operations(self):

        if not self.__check_array():
            return

        self.print_line()
        print("Mathematical Operations")
        self.print_line()

        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        choice = input("Enter your choice: ")

        if choice in ["1", "2", "3", "4"]:

            print("\nOriginal Array:")
            print(self.__array)

            print("\nEnter",self.__array.size,"elements for second array:")

            try:

                values = list(map(float, input().split()))

                if len(values) != self.__array.size:
                    print("Invalid number of elements.")
                    return

                second = np.array(values).reshape(self.__array.shape)

                print("\nSecond Array:")
                print(second)

                if choice == "1":

                    result = self.__array + second

                    print("\nResult of Addition:")
                    print(result)

                elif choice == "2":

                    result = self.__array - second

                    print("\nResult of Subtraction:")
                    print(result)

                elif choice == "3":

                    result = self.__array * second

                    print("\nResult of Multiplication:")
                    print(result)

                elif choice == "4":

                    if np.any(second == 0):
                        print("Division by zero is not allowed.")
                        return

                    result = self.__array / second

                    print("\nResult of Division:")
                    print(result)

            except ValueError:
                print("Invalid input.")

        elif choice == "5":

            if self.__array.ndim != 2:
                print("Dot product requires a 2D array.")
                return

            rows, columns = self.__array.shape

            print("Enter",rows * columns,"elements for second array:")

            try:

                values = list(map(float, input().split()))

                if len(values) != rows * columns:
                    print("Invalid number of elements.")
                    return

                second = np.array(values).reshape(rows, columns)

                print("\nSecond Array:")
                print(second)

                print("\nDot Product:")
                print(np.dot(self.__array, second))

            except ValueError:
                print("Invalid input.")

        elif choice == "6":

            if self.__array.ndim != 2:
                print("Matrix multiplication requires a 2D array.")
                return

            rows = self.__array.shape[1]

            try:

                columns = int(input("Enter number of columns of second matrix: "))

                total = rows * columns

                print("Enter",total,"elements for second matrix:")

                values = list(map(float, input().split()))

                if len(values) != total:
                    print("Invalid number of elements.")
                    return

                second = np.array(values).reshape(rows, columns)

                print("\nSecond Matrix:")
                print(second)

                print("\nMatrix Multiplication:")
                print(np.matmul(self.__array,second))

            except ValueError:
                print("Invalid input.")

        else:
            print("Invalid choice.")

    
    def combine_split(self):

        if not self.__check_array():
            return

        self.print_line()
        print("Combine or Split Arrays")
        self.print_line()

        print("1. Combine Arrays")
        print("2. Split Array")

        try:

            choice = int(input("Enter your choice: "))

            if choice == 1:

                print("Original Array:")
                print(self.__array)

                print("Enter",self.__array.size,"elements for second array:")

                values = list(map(float, input().split()))

                if len(values) != self.__array.size:
                    print("Invalid number of elements.")
                    return

                second = np.array(values).reshape(self.__array.shape)

                print("\nSecond Array:")
                print(second)

                try:

                    result = np.concatenate((self.__array, second),axis=0)

                    print("\nCombined Array:")
                    print(result)

                except ValueError:
                    print("Arrays cannot be combined with these dimensions.")

            elif choice == 2:

                parts = int(input("Enter number of parts: "))

                if parts <= 0:
                    print("Number of parts must be greater than 0.")
                    return

                result = np.array_split(self.__array,parts)

                print("\nSplit Arrays:")

                for i in range(len(result)):

                    print("\nPart", i + 1, ":")
                    print(result[i])

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter valid numeric values.")

    
    def search_short_filter(self):

        if not self.__check_array():
            return

        self.print_line()
        print("Search, Sort, Filter")
        self.print_line()

        print("1. Search a value")
        print("2. Sort the array")
        print("3. Filter values")

        try:

            choice = int(input("Enter your choice: "))

            if choice == 1:

                value = float(input("Enter value to search: "))
                positions = np.where(self.__array == value)

                if len(positions[0]) == 0:
                    print("Value not found.")

                else:
                    print("Value found at:")
                    print(positions)

            elif choice == 2:

                print("1. Ascending")
                print("2. Descending")

                sort_choice = int(input("Enter your choice: "))

                sorted_array = np.sort(self.__array.flatten())

                if sort_choice == 1:

                    print("Ascending Order:")
                    print(sorted_array)

                elif sort_choice == 2:

                    print("Descending Order:")
                    print(sorted_array[::-1])

                else:
                    print("Invalid choice.")

            elif choice == 3:

                print("1. Greater than")
                print("2. Less than")
                print("3. Equal to")

                filter_choice = int(input("Enter your choice: "))

                value = float(input("Enter value: "))

                if filter_choice == 1:

                    result = self.__array[self.__array > value]

                elif filter_choice == 2:

                    result = self.__array[self.__array < value]

                elif filter_choice == 3:

                    result = self.__array[self.__array == value]

                else:
                    print("Invalid choice.")
                    return

                print("Filtered values:")
                print(result)

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter valid numeric values.")

    
    def statistics(self):

        if not self.__check_array():
            return

        self.print_line()
        print("Aggregates and Statistics")
        self.print_line()

        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum")
        print("7. Maximum")
        print("8. Percentile")
        print("9. Correlation Coefficient")

        try:

            choice = int(input("Enter your choice: "))

            if choice == 1:

                print("\nSum:",np.sum(self.__array))

            elif choice == 2:

                print("\nMean:",np.mean(self.__array))

            elif choice == 3:

                print("\nMedian:",np.median(self.__array))

            elif choice == 4:

                print("\nStandard Deviation:",np.std(self.__array))

            elif choice == 5:

                print("\nVariance:",np.var(self.__array))

            elif choice == 6:

                print("\nMinimum:",np.min(self.__array))

            elif choice == 7:

                print("\nMaximum:",np.max(self.__array))

            elif choice == 8:

                value = float(input("Enter percentile (0-100): "))

                if value < 0 or value > 100:

                    print("Percentile must be between 0 and 100.")

                else:

                    print("\nPercentile:",np.percentile(self.__array,value))

            elif choice == 9:

                print("Enter",self.__array.size,"values for second array:")

                values = list(map(float, input().split()))

                if len(values) != self.__array.size:

                    print("Invalid number of elements.")
                    return

                second = np.array(values)
                first = self.__array.flatten()

                correlation = np.corrcoef(first,second)[0, 1]

                print("\nCorrelation Coefficient:",correlation)

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter valid numeric values.")



def main():

    analyzer = DataAnalyticcs.create_analyzer()

    while True:

        DataAnalyticcs.print_line()
        print("Welcome to Numpy Analyzer!")
        DataAnalyticcs.print_line()

        print("Choose an option:")

        print("1. Create a Numpy Array")
        print("2. Indexing and Slicing")
        print("3. Perform Mathematical Operations")
        print("4. Combine or Split Arrays")
        print("5. Search, Sort, or Filter Arrays")
        print("6. Compute Aggregates and Statistics")
        print("7. Exit")

        try:

            choice = int(input("Enter your choice: "))

            if choice == 1:

                analyzer.create_array()

            elif choice == 2:

                analyzer.indexing_slicing()

            elif choice == 3:

                analyzer.mathematical_operations()

            elif choice == 4:

                analyzer.combine_split()

            elif choice == 5:

                analyzer.search_short_filter()

            elif choice == 6:

                analyzer.statistics()

            elif choice == 7:

                print( "Thank you for using NumPy Analyzer!")
                break

            else:

                print("Invalid choice.")

        except ValueError:

            print("Please enter a valid number.")


if __name__ == "__main__":
    main()
```
