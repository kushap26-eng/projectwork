```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

class SalesDataAnalyzer:
    def __init__(self):
        self.df = None
        self.figured = {}
        print("Google Ads Sales Data Analyzer initialized.")

    def clean_and_convert_column(self, series):
        """Helper function to strip currency symbols and commas before converting to float."""
        cleaned = series.astype(str).str.replace('$', '', regex=False)
        cleaned = cleaned.str.replace(',', '', regex=False)
        return pd.to_numeric(cleaned, errors="coerce")

    def load_data(self):
        filepath = input("Enter CSV/Excel file path: ")
        if not filepath:
            filepath = "google_ads_sales.csv"
        try:
            if filepath.lower().endswith(".csv"):
                self.df = pd.read_csv(filepath)
            elif filepath.lower().endswith((".xlsx", '.xls')):
                self.df = pd.read_excel(filepath)
            else:
                print("Unsupported file format. Use CSV or Excel.")
                return
            print("Dataset loaded successfully!")
            print("Number of rows: ", self.df.shape[0])
            print("Number of columns: ", self.df.shape[1])
            print("Column names: ", self.df.columns.tolist())
        except FileNotFoundError:
            print("File not found. Check the file path and try again.")
        except Exception as error:
            print("Error loading dataset: ", error)

    def check_data(self):
        if self.df is None:
            print("Please load a dataset first.")
            return False
        return True

    def numeric_data(self):
        return self.df.select_dtypes(include=[np.number])

    def explore_data(self):
        if not self.check_data():
            return
        while True:
            print("---------- Explore Data ----------")
            print("1. Display first 5 rows")
            print("2. Display last 5 rows")
            print("3. Display column names")
            print("4. Display data types")
            print("5. Display basic information")
            print("6. Display dataset shape")
            print("7. Back to main menu")
            choice = input("Enter your choice: ")
            
            if choice == "1":
                print(self.df.head())
            elif choice == "2":
                print(self.df.tail())
            elif choice == "3":
                print(self.df.columns.tolist())
            elif choice == "4":
                print(self.df.dtypes)
            elif choice == "5":
                self.df.info()
            elif choice == "6":
                print("Rows and columns:", self.df.shape)
            elif choice == "7":
                break
            else:
                print("Invalid choice. Try again")

    def dataframe_operations(self):
        if not self.check_data():
            return
        while True:
            print("------- DataFrame and NumPy Operations -------")
            print("1. Display DataFrame shape, columns and types")
            print("2. Select specific columns")
            print("3. NumPy mathematical operations")
            print("4. Search, sort and filter data")
            print("5. Group data and aggregate")
            print("6. Create pivot table")
            print("7. Advanced operations")
            print("8. Back to main menu")
            choice = input("Enter your choice: ")
            
            if choice == "1":
                print("Shape: ", self.df.shape)
                print("Columns: ", self.df.columns.tolist())
                print("Data types: ", self.df.dtypes)
            elif choice == "2":
                print("Available columns:", self.df.columns.tolist())
                columns = input("Enter column names separated by commas: ")
                selected = [column.strip() for column in columns.split(",")]
                try:
                    print(self.df[selected].head())
                except KeyError as error:
                    print("Invalid column name:", error)
            elif choice == "3":
                self.numpy_operations()
            elif choice == "4":
                self.search_sort_filter()
            elif choice == "5":
                self.aggregate_data()
            elif choice == "6":
                self.create_pivot_table()
            elif choice == "7":
                self.advanced_operations()
            elif choice == "8":
                break
            else:
                print("Invalid choice. Try again.")

    def numpy_operations(self):
        numeric_df = self.numeric_data()
        if numeric_df.empty:
            print("No numeric columns are available.")
            return
        
        array = numeric_df.dropna().to_numpy()
        if array.size == 0:
            print("Numeric columns contain only missing values.")
            return

        print("------ NumPy Operations ------")
        print("Array shape:", array.shape)
        print("Array size:", array.size)
        print("Sum:", np.sum(array))
        print("Mean:", np.mean(array))
        print("Standard deviation:", np.std(array))
        print("Variance:", np.var(array))
        print("Minimum:", np.min(array))
        print("Maximum:", np.max(array))
        print("Median:", np.median(array))
        
        if array.shape[0] > 0:
            print("First row:", array[0])
            print("First three rows and first two columns: ")
            print(array[:3, :2])
            
        flat_array = array.flatten()
        print("First 10 sorted values:", np.sort(flat_array)[:10])
        
        threshold = input("Enter a threshold to find values greater than it: ")
        try:
            threshold = float(threshold)
        except ValueError:
            print("Invalid threshold. Using 100.")
            threshold = 100.0
            
        indices = np.where(flat_array > threshold)[0]
        print(f"Indices with values above {threshold}:", indices[:20])

    def search_sort_filter(self):
        if not self.check_data():
            return
        print("------ Search, Sort and Filter ------")
        print("1. Filter rows by exact value")
        print("2. Sort by a column")
        print("3. Search text in a column")
        print("4. Filter numeric values above threshold")
        choice = input("Enter your choice: ")
        
        if choice == "1":
            column = input("Column Name: ")
            value = input("Value to match: ").strip().lower() 
            if column in self.df.columns:
                mask = self.df[column].astype(str).str.strip().str.lower() == value
                filtered_df = self.df[mask]
                
                if filtered_df.empty:
                    print("No exact matches found. Check your spelling or character spacing.")
                else:
                    print(filtered_df.head(20))
            else:
                print("Column not found.")

        elif choice == "2":
            column = input("Column to sort by: ")
            if column in self.df.columns:
                ascending = input("Ascending order? (y/n): ").strip().lower() == "y"
                print(self.df.sort_values(column, ascending=ascending).head(20))
            else:
                print("Column not found.")
        elif choice == "3":
            column = input("Text column name: ")
            if column in self.df.columns:
                search_text = input("Text to search: ")
                mask = self.df[column].astype(str).str.contains(search_text, case=False, na=False)
                print(self.df[mask].head(20))
            else:
                print("Column not found.")
        
        elif choice == "4":
            column = input("Numeric column name: ")
            if column in self.df.columns:
                try:
                    threshold = float(input("Enter threshold: "))
                    cleaned_series = self.df[column].astype(str).str.replace('$', '', regex=False)
                    cleaned_series = cleaned_series.str.replace(',', '', regex=False)
                    cleaned_series = cleaned_series.str.replace('%', '', regex=False).str.strip()
                    
                    values = pd.to_numeric(cleaned_series, errors="coerce")
                    
                    filtered_df = self.df[values > threshold]
                    
                    if filtered_df.empty:
                        print(f"No records found with values greater than {threshold} in column '{column}'.")
                    else:
                        print(filtered_df.head(20))
                        
                except ValueError:
                    print("Please enter a valid number for the threshold.")
            else:
                print("Column not found.")

        else:
            print("Invalid choice.")

    def aggregate_data(self):
        if not self.check_data():
            return
        print("------ Aggregate Data ------ ")
        print("Available columns:", self.df.columns.tolist())
        group_column = input("Group by column: ")
        value_column = input("Numeric column to summarize: ")
        
        if group_column not in self.df.columns or value_column not in self.df.columns:
            print("Please enter valid column names.")
            return
            
        values = self.clean_and_convert_column(self.df[value_column])
        if values.notna().sum() == 0:
            print("Selected value column has no valid numeric values.")
            return
            
        print("1. Sum\n2. Mean\n3. Count\n4. Minimum\n5. Maximum")
        choice = input("Choose aggregation: ")
        operations = {"1": "sum", "2": "mean", "3": "count", "4": "min", "5": "max"}
        
        if choice not in operations:
            print("Invalid aggregation choice.")
            return
            
        temp = self.df[[group_column]].copy()
        temp[value_column] = values
        print(temp.groupby(group_column)[value_column].agg(operations[choice]))

    def create_pivot_table(self):
        if not self.check_data():
                return
        print("------ Create Pivot Table ------")
        index_column = input("Index/group column: ")
        columns_column = input("Optional columns column (press enter to skip): ")
        values_column = input("Values column: ")
        function = input("Aggregation (sum/mean/count) [sum]: ").strip().lower() or "sum"
        
        if function not in ["sum", "mean", "count"]:
            print("Choose sum, mean or count.")
            return
        if index_column not in self.df.columns or values_column not in self.df.columns:
            print("Index or values column not found.")
            return
        if columns_column and columns_column not in self.df.columns:
            print("Columns field not found.")
            return
            
        data = self.df.copy()
        
        
        data[index_column] = data[index_column].astype(str).str.strip()
        if columns_column:
            data[columns_column] = data[columns_column].astype(str).str.strip()
            
        if function in ["sum", "mean"]:
            cleaned_values = data[values_column].astype(str).str.replace('$', '', regex=False)
            cleaned_values = cleaned_values.str.replace(',', '', regex=False).str.strip()
            data[values_column] = pd.to_numeric(cleaned_values, errors="coerce")
            
        try:
            result = pd.pivot_table(data, index=index_column, columns=columns_column if columns_column else None, values=values_column, aggfunc=function)
            
            if columns_column:
                result = result.reset_index()
                result.columns.name = None  
            else:
                result = result.reset_index()
            print(result.to_string(index=False)) 
              
        except Exception as error:
            print("Could not create pivot table:", error)

    def advanced_operations(self):
        if not self.check_data():
            return
        print("------ Advanced Operations ------")
        print("1. Re-index by column")
        print("2. Groupby transform (group mean)")
        print("3. Add Google Ads performance metrics")
        choice = input("Enter your choice: ")   

        if choice == "1":
            column = input("Column to use as index: ")
            if column in self.df.columns:
                self.df = self.df.set_index(column)
                print(self.df.head())
            else:
                print("Column not found.")    
        
        elif choice == "2":
            group_column = input("Group by column: ")
            numeric_column = input("Numeric column: ")
            
            if group_column not in self.df.columns or numeric_column not in self.df.columns:
                print("Column not found.")
                return
    
            cleaned_series = self.df[numeric_column].astype(str).str.replace('$', '', regex=False)
            cleaned_series = cleaned_series.str.replace(',', '', regex=False).str.strip()
           
            self.df[numeric_column] = pd.to_numeric(cleaned_series, errors="coerce")
            new_column = numeric_column + "_group_mean"
            self.df[new_column] = self.df.groupby(group_column)[numeric_column].transform("mean")
            
            print("\nTransformation Completed Successfully:")
            print(self.df[[group_column, numeric_column, new_column]].head())

        elif choice == "3":
            self.add_ads_metrics()
        
        else:
            print("Invalid choice.")

    def add_ads_metrics(self):
        lower_names = {str(column).lower().strip(): column for column in self.df.columns}
        
        def find_column(possible_names):
            for name in possible_names:
                if name in lower_names:
                    return lower_names[name]
            return None

        def safe_numeric_clean(col_name):
            if col_name is None:
                return None
            cleaned = self.df[col_name].astype(str).str.replace('$', '', regex=False)
            cleaned = cleaned.str.replace(',', '', regex=False).str.strip()
            return pd.to_numeric(cleaned, errors="coerce")

        clicks = find_column(["clicks", "click"])
        cost = find_column(["cost", "spend", "ad_spend", "amount_spent"])
        conversions = find_column(["conversions", "conversion"])
        revenue = find_column(["revenue", "conversion_value", "sales", "sale_amount", "sale"])

        clk_series = safe_numeric_clean(clicks)
        spend_series = safe_numeric_clean(cost)
        conv_series = safe_numeric_clean(conversions)
        rev_series = safe_numeric_clean(revenue)

        if clk_series is not None and spend_series is not None:
            self.df[clicks] = clk_series
            self.df[cost] = spend_series
            self.df["CPC"] = np.where(clk_series > 0, spend_series / clk_series, 0.0)
            print("CPC column added successfully.")
        else:
            print("CPC not calculated: matching Clicks/Cost columns not found.")

        if clk_series is not None and conv_series is not None:
            self.df[conversions] = conv_series
            self.df["Calculated_CR"] = np.where(clk_series > 0, (conv_series / clk_series) * 100, 0.0)
            print("Calculated_CR column added successfully.")
        else:
            print("Conversion rate not calculated: matching Clicks/Conversions columns not found.")

        if spend_series is not None and rev_series is not None:
            self.df[revenue] = rev_series
            self.df["ROAS"] = np.where(spend_series > 0, rev_series / spend_series, 0.0)
            print("ROAS column added successfully.")
        else:
            print("ROAS not calculated: matching Revenue/Cost columns not found.")

        print(self.df.head())

    def handle_missing_data(self):
        if not self.check_data():
            return
        while True:
            print("\n------ Handle Missing Data ------")
            print("1. Display rows with missing values")
            print("2. Fill numeric missing values with mean")
            print("3. Replace missing values with a specific value")
            print("4. Drop rows with missing values")
            print("5. Back to main menu")
            choice = input("Enter your choice: ")
            
            string_check = self.df.astype(str).apply(lambda x: x.str.strip().str.lower())
            missing_rows_mask = (self.df.isna()) | (string_check == "nan") | (string_check == "")
            
            if choice == "1":
                missing_data = self.df[missing_rows_mask.any(axis=1)]
                if not missing_data.empty:
                    print("\nHere are your rows with missing values:")
                    print(missing_data)
                else:
                    print("No missing values found.")
                    
            elif choice == "2":
                for col in self.df.columns:
                    if col in self.numeric_data().columns:
                        self.df.loc[missing_rows_mask[col], col] = np.nan
                        self.df[col] = pd.to_numeric(self.df[col], errors='coerce')
                        self.df[col] = self.df[col].fillna(self.df[col].mean())
                print("Missing numbers filled with column averages successfully!")
                
            elif choice == "3":
                fill_value = input("Enter replacement value (e.g. 0): ")
                for col in self.df.columns:
                    self.df.loc[missing_rows_mask[col], col] = fill_value
                print(f"All missing values replaced with: {fill_value}")
                
            elif choice == "4":
                before = len(self.df)
                self.df = self.df[~missing_rows_mask.any(axis=1)].copy()
                print(f"Cleaned up! Removed {before - len(self.df)} rows from your dataset.")
                
            elif choice == "5":
                break
            else:
                print("Invalid choice. Try again.")


    def statistical_analysis(self):
        if not self.check_data(): 
            return
            
        cleaned_df = self.df.copy()
        for col in cleaned_df.columns:
            if cleaned_df[col].dtype == 'object':
                cleaned = cleaned_df[col].astype(str).str.replace('$', '', regex=False)
                cleaned = cleaned.str.replace(',', '', regex=False).str.strip()
                numeric_col = pd.to_numeric(cleaned, errors='coerce')
                if numeric_col.notna().sum() > 0:
                    cleaned_df[col] = numeric_col
        numeric_df = cleaned_df.select_dtypes(include=[np.number])
        
        if numeric_df.empty:
            print("No numeric columns available for statistical analysis.")
            return

        print("------ Descriptive Statistics ------")
        print(numeric_df.describe())
        print("Variance:", numeric_df.var())
        print("Standard Deviation:", numeric_df.std())
        print("Skewness:", numeric_df.skew())
        print("Kurtosis:", numeric_df.kurtosis())
        print("Percentiles (25th, 50th, 75th):", numeric_df.quantile([0.25, 0.50, 0.75]))

    def visualize_data(self):
        if not self.check_data():
            return
        
        while True:
            print("\n------ Data Visualization ------")
            print("1. Bar plot")
            print("2. Line plot")
            print("3. Scatter plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack plot")
            print("7. Seaborn heatmap")
            print("8. Seaborn box plot")
            print("9. Back to main menu")
            
            choice = input("Enter your choice: ")
            if choice == "9":
                break
                
            plt.figure()

            def get_clean_numeric(col_name):
                cleaned = self.df[col_name].astype(str).str.replace('$', '', regex=False)
                cleaned = cleaned.str.replace(',', '', regex=False).str.strip()
                return pd.to_numeric(cleaned, errors='coerce')

            try:
                if choice == "1":
                    x = input("Enter category column: ")
                    y = input("Enter numeric column: ")
                    
                    temp = self.df.copy()
                    temp[y] = get_clean_numeric(y)
                    
                    data = temp.groupby(x)[y].sum().dropna()
                    data.plot(kind="bar")
                    plt.title(f"Total {y} by {x}")
                    plt.xlabel(x)
                    plt.ylabel(y)
                    plt.xticks(rotation=45)
                    plt.tight_layout()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "2":
                    x = input("Enter X-axis/Date column: ")
                    y = input("Enter numeric column: ")
                    
                    temp = self.df.copy()
                    temp[y] = get_clean_numeric(y)
                    
                    if "date" in x.lower():
                        temp[x] = pd.to_datetime(temp[x], errors='coerce', dayfirst=True)
                    
                    temp = temp.dropna(subset=[x, y])
                    trend_data = temp.groupby(x)[y].sum().sort_index()

                    plt.plot(trend_data.index, trend_data.values, marker="o", linestyle="-", linewidth=2)
                    plt.title(f"Total {y} Trend over {x}")
                    plt.xlabel(x)
                    plt.ylabel(f"Total {y}")
                    plt.xticks(rotation=45)
                    plt.tight_layout()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "3":
                    x = input("Enter X-axis numeric column: ")
                    y = input("Enter y-axis numeric column: ")
                    
                    plt.scatter(get_clean_numeric(x), get_clean_numeric(y))
                    plt.title(f"{y} vs {x}")
                    plt.xlabel(x)
                    plt.ylabel(y)
                    plt.tight_layout()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "4":
                    category = input("Enter category column: ")
                    value = input("Enter numeric value column: ")
                    
                    temp = self.df.copy()
                    temp[value] = get_clean_numeric(value)
                    
                    data = temp.groupby(category)[value].sum().dropna()
                    data = data[data > 0].head(10)
                    if data.empty:
                        print("No positive numeric values available to build chart.")
                        continue
                    data.plot(kind="pie", autopct="%1.1f%%")
                    plt.ylabel("")
                    plt.title(f"{value} Distribution by {category}")
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "5":
                    column = input("Enter numeric column: ")
                    values = get_clean_numeric(column).dropna()
                    plt.hist(values, bins=10, edgecolor="black")
                    plt.title(f"Histogram of {column}")
                    plt.xlabel(column)
                    plt.ylabel("Frequency")
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "6":
                    cleaned_df = pd.DataFrame()
                    for col in self.df.columns:
                        clean_col = get_clean_numeric(col)
                        if clean_col.notna().sum() > 0:
                            cleaned_df[col] = clean_col.fillna(0)
                            
                    if cleaned_df.shape[1] < 2:
                        print("At least two valid numeric columns are required to draw a Stack Plot.")
                        continue
                    data = cleaned_df.iloc[:, :4]
                    plt.stackplot(range(len(data)), *[data[col] for col in data.columns], labels=data.columns)
                    plt.title("Stack Plot")
                    plt.xlabel("Record Number")
                    plt.ylabel("Values")
                    plt.legend()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "7":
                    cleaned_df = pd.DataFrame()
                    for col in self.df.columns:
                        clean_col = get_clean_numeric(col)
                        if clean_col.notna().sum() > 0:
                            cleaned_df[col] = clean_col
                            
                    if cleaned_df.shape[1] < 2:
                        print("At least two valid numeric columns are required to measure a Correlation Heatmap.")
                        continue
                    sns.heatmap(cleaned_df.corr(), annot=True)
                    plt.title("Correlation Heatmap")
                    plt.tight_layout()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                elif choice == "8":
                    column = input("Enter numeric column: ")
                    values = get_clean_numeric(column).dropna()
                    if values.empty:
                        print("No valid numeric values available to build box chart.")
                        continue
                    sns.boxplot(x=values)
                    plt.title(f"Box Plot of {column}")
                    plt.xlabel(column)
                    plt.tight_layout()
                    self.figured["latest"] = plt.gcf()
                    plt.show()

                else:
                    print("Invalid choice. Please select a valid number from 1 to 9.")
                    continue
                print("Visualization created successfully.")
            except KeyError:
                print("Invalid column name. Check your spelling and spacing.")
            except (ValueError, TypeError) as error:
                print("Unable to create visualization: ", error)

    def run_menu(self):
        while True:
            print("\n" + "-"*30)
            print(" Google Ads Data Analysis & Visualization")
            print("="*30)
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. Perform DataFrame Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Exit")
            choice = input("Enter your choice: ")
            
            if choice == "1":
                self.load_data()
            
            elif choice == "2":
                self.explore_data()
            
            elif choice == "3":
                self.dataframe_operations()
                
            elif choice == "4":
                self.handle_missing_data()
                
            elif choice == "5":
                self.statistical_analysis()
                
            elif choice == "6":
                self.visualize_data()
                
            elif choice == "7":
                self.save_visualization()
                
            elif choice == "8":
                print("Exiting the program. Thank You!")
                break
            
            else:
                print("Invalid choice. Please enter a number from 1 to 8.")

if __name__ == "__main__":
    analyzer = SalesDataAnalyzer()
    analyzer.run_menu()
```
